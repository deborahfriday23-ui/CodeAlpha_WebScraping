```python
def extract_book(item):
    title = item.select_one("h3 a")["title"]

    price = item.select_one(".price_color").get_text(strip=True)
    price = price.replace("Â", "")

    rating = item.select_one("p.star-rating")["class"][1]
    availability = item.select_one(".availability").get_text(strip=True)

    book = {
        "Title": title,
        "Price": price,
        "Rating": rating,
        "Availability": availability
    }

    return book
```

```python
all_books_data = []

for page_number in range(1, 6):
    url = f"https://books.toscrape.com/catalogue/page-{page_number}.html"

    response = requests.get(url)
    page_soup = BeautifulSoup(response.text, "html.parser")

    books_on_page = page_soup.select("article.product_pod")

    for item in books_on_page:
        book = extract_book(item)
        all_books_data.append(book)

print("Total books collected:", len(all_books_data))
```

```python
import time
```

```python
all_books_data = []

for page_number in range(1, 51):
    url = f"https://books.toscrape.com/catalogue/page-{page_number}.html"

    response = requests.get(url, timeout=15)

    if response.status_code == 200:
        page_soup = BeautifulSoup(response.text, "html.parser")
        books_on_page = page_soup.select("article.product_pod")

        for item in books_on_page:
            book = extract_book(item)
            all_books_data.append(book)

        print(f"Page {page_number}: collected {len(books_on_page)} books")
    else:
        print(f"Page {page_number}: failed with status {response.status_code}")

    time.sleep(1)

print("Total books collected:", len(all_books_data))
```

```python
import pandas as pd

df = pd.DataFrame(all_books_data)

print("Dataset shape:", df.shape)
df.head()
```

```python
# Remove the pound symbol from prices
df["Price"] = df["Price"].str.replace("£", "", regex=False)

# Convert prices from text to numbers
df["Price"] = df["Price"].astype(float)

# Check the result
print(df["Price"].head())
print("\nPrice data type:", df["Price"].dtype)
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate book titles:")
print(df["Title"].duplicated().sum())
```

```python
duplicate_books = df[df["Title"].duplicated(keep=False)]

print(duplicate_books)
```

```python
print("Average price: £", round(df["Price"].mean(), 2))
print("Cheapest price: £", df["Price"].min())
print("Most expensive price: £", df["Price"].max())
```

```python
rating_counts = df["Rating"].value_counts()

print(rating_counts)
```
